# **🏷️ 1️⃣ Introduction to Control Structures & Looping Structures**

# 🚦 Control Structures & Looping Structures

Control structures in R dictate the flow of execution of your code. They allow you to make decisions, repeat actions, and generally control 'what happens when'.

The primary control structures we will cover are:
- `if`, `else if`, `else` (Conditional statements)
- `for` (Loops)
- `while` (Loops)

# **🏷️ 2️⃣ Conditional Statements: if, else if, else**

# ❓ if-else Statements

Conditional statements execute different blocks of code based on whether a condition is `TRUE` or `FALSE`.

Syntax:
```R
if (condition) {
  # code to execute if condition is TRUE
} else if (another_condition) {
  # code to execute if another_condition is TRUE
} else {
  # code to execute if all conditions are FALSE
}
```
Examples:

````r
# Simple if statement
temperature <- 28
if (temperature > 25)
{
  print("It's warm.")
}
````

````r
# if-else statement
# In R, else must appear on the same line as the closing } of the preceding block.
age <- 20
if (age >= 18)
{
  print("You are an adult.")
} else {
  print("You are a minor.")
}
````

````r
# if - else if - else
age <- 4
if (age == 18)
{
  print("You are 18 years old.")
} else if (age > 18) {
  print("You are an adult.")
} else {
  print("You are a minor")
}
````

# **🏷️ 3️⃣ Loops: for Loop**

# 🔁 for Loop

The `for` loop iterates over a sequence (like a vector or list) or a specified range of numbers.

Syntax:
```R
for (variable in sequence) {
  # code to execute for each item in sequence
}
```
Examples:

````r
# Loop through a numeric vector
numbers <- c(1, 2, 3, 4, 5)
for (num in numbers) {
  print(num * 2)
}
````

````r
# Loop through a character vector
fruits <- c("apple", "banana", "cherry")
for (fruit in fruits) {
  print(paste("I like", fruit))
}
````

````r
# Loop using a sequence (from 1 to 3)
for (i in 1:3) {
  print(paste("Iteration", i))
}
````

# **🏷️ 4️⃣ Loops: while Loop**

# 🔄 while Loop

The `while` loop repeatedly executes a block of code as long as a specified condition is `TRUE`.

Syntax:
```R
while (condition) {
  # code to execute as long as condition is TRUE
}
```
Examples:

````r
count <- 1
while (count <= 5) {
  print(paste("Count is:", count))
  count <- count + 1
}
````

````r
# Example with breaking a loop
i <- 1
while (TRUE) {
  print(paste("Current i is:", i))
  if (i >= 3) {
    break # Exit the loop
  }
  i <- i + 1
}
````

# **🏷️ 6️⃣ Quick Quiz**

# ✅ Quick Quiz

1️⃣ When would you use an `if` statement?  
'if' is used when we need a deision for a condition, example: print TRUE, if x=5.

2️⃣ What's the main difference between a `for` loop and a `while` loop?  
->1. 'for' loop is used when you know how many times the loop should run, on the other hand 'while' loop is used when we don't know how times the loop should run.
->2. 'for' loop runs over a sequence (vector, list, range), whereas, 'while' loop runs as long as the condition is TRUE.
->3.'for' loop is easier to control, and 'while' loop needs careful condition control to avoid infinite loops.

3️⃣ How do you exit a `while` loop prematurely in R?  
We use the break statement to stop a loop immediately.
